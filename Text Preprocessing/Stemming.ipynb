```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = "walk walks walking walked"
stem_words(sample)
```

```python
text = "The organization is organizing an organizational meeting to organize future organizers. Connecting connected connections allows reconnectivity across nodes."
print(text)
```

```python
stem_words(text)
```
