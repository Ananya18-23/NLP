```python
chat_words = {
    "im": "i am",
    "r": "are",
    "u": "you",
    "brb": "be right back",
    "btw": "by the way",
    "fyi": "for your information",
    "idk": "i do not know",
    "omg": "oh my god",
    "ttyl": "talk to you later",
    "gn": "good night",
    "tue": "tuesday"
}
```

```python
# 2. Function jo chat words ko replace karega
def convert_chat_words(text):
    new_text = []
    for word in text.split():
        # Lowercase karke check karein taaki matching sahi ho
        if word.lower() in chat_words:
            new_text.append(chat_words[word.lower()])
        else:
            new_text.append(word)
    return " ".join(new_text)
```

```python
convert_chat_words("I can't believe it! ttyl.")
```
