### 1 using the split function

```python
# word tenization
sent1 = 'i am going to the market'
sent1.split()
```

```python
# sentence tokenization
sent2 = 'i am going to the market. i will buy some vegetables. i will also buy some fruits.'
sent2.split('.')
```

```python
# problem with split function
sent3 = 'i am going to bhoapl!'
sent3.split()
```

### 2 Regular Expression

```python
import re
sent3 = 'i am going to bhoapl!'
tokens = re .findall("[\w']+", sent3)
tokens
```

```python
text = """ lorem ipsum is simply dummy text of the printing and typesetting industry"""
sentences = re.compile('[.!?]').split(text)
sentences
```

### 3 NLTK

```python
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
sent1 = 'i am going to the market. i will buy some vegetables. i will also buy some fruits.'
word_tokenize(sent1)
```

```python
text =""" lorem ipsum is simply dummy text of the printing and typesetting industry"""
sent_tokenize(text)
```

```python
sent5 = ' I have a ph.D in A.I'
sent6 = "we're here to help! mail us at nks@gmail.com"
sent7 = 'A 5km ride costs $10.50'
```

```python
word_tokenize(sent5)
```

```python
word_tokenize(sent6)
```

```python
word_tokenize(sent7)
```
